```python
import os
from datetime import datetime

def archive_file(source_path: str, archive_dir: str) -> str:
    """
    Moves a processed file to an archive location with a timestamp prefix.
    """
    try:
        dbutils.fs.ls(source_path)
    except Exception:
        raise FileNotFoundError(f"Source file not found at: {source_path}")

    # Extract file name and build timestamped target path
    base_name = os.path.basename(source_path)
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    
    archive_dir = archive_dir.rstrip("/")
    target_path = f"{archive_dir}/{timestamp}_{base_name}"

    # Ensure archive folder exists
    #dbutils.fs.mkdirs(archive_dir)

    # Move file
    dbutils.fs.mv(source_path, target_path)
    print(f"Archived: {source_path} -> {target_path}")
    return target_path


# ---  Execution ---
source_file = "/Volumes/workspace/shprod/payrolldata/data/payrolldata.csv"
archive_dir = "/Volumes/workspace/shprod/payrolldata/archive/nycpayrolldata"

archive_file(source_file, archive_dir)
```
